## Topic 7: Functions & Passing Arguments

### Q7.1
Define `my_abs(val)` without using built-in `abs()`.

```python
def my_abs(val):
    if val < 0:
        return 0 - val
    return val

print(my_abs(7))
print(my_abs(-7))
print(my_abs(0))
```

### Q7.2
What happens with `my_abs("Hi")`?

**Answer:** It raises `TypeError` because the function tries to evaluate `"Hi" < 0`, and Python cannot order a string and an integer.

```python
try:
    print(my_abs("Hi"))
except TypeError as e:
    print("TypeError:", e)
```

### Q7.3
A function that prints rather than returns the absolute value.

```python
def print_abs(val):
    if val < 0:
        print(0 - val)
    else:
        print(val)

x = print_abs(-2.7)
print("x =", x)
```

**Answer:** `2.7` is printed by the function, then `x` is `None` because the function has no `return` statement.

### Q7.4
Predict and verify.

**Output:** `7`

**Reason:** The integer object referenced by `x` is not changed. The local name `val` is reassigned inside the function, and that local reassignment does not change `x`.

## Topic 8: Scope — Local vs Global Variables

### Q8.1
Predict and verify the scope error.

```python
def my_abs(val):
    if val < 0:
        return 0 - val
    return val

try:
    print(val)
except NameError as e:
    print("NameError:", e)
```

**Reason:** `val` is a local variable belonging to the function. It does not exist in the global scope after the function definition.

### Q8.2
Read a global variable from inside a function.

```python
counter = 0

def show_counter():
    print(counter)

show_counter()
```

### Q8.3
Try to increment a global variable without `global`.

```python
counter = 0

def increment_counter():
    try:
        counter = counter + 1
    except UnboundLocalError as e:
        print("UnboundLocalError:", e)

increment_counter()
```

**Reason:** Because `counter` is assigned inside the function, Python treats it as a local variable. Reading it before that local assignment causes `UnboundLocalError`.

### Q8.4
Fix it with the `global` keyword.

```python
counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)
```
